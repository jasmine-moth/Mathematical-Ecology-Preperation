# Stability and Cycles

----

In the previous sections, we have discovered 2 different ecological models, with those being the logistic equation and the Lotka-Volterra model. Now we turn our attention to developing the rigour behind understanding stability and equilibrium points. 

Firstly, let us consider an autonomous predator-prey system:

$$
\frac{dN}{dt} = F(N,P) \tag{3.1a}
$$

$$
\frac{dP}{dt} = G(N,P) \tag{3.1b}
$$

Where $N$ is the population of prey, and $P$ is the predator population. We call them autonomous because the functions $F$ and $G$ have no explicit dependence on time. From earlier, we know that $(N^{*},P^{*})$ is an equilibrium point if:

$$
F(N^{*},P^{*}) = 0 \quad and \quad G(N^{*},P^{*}) = 0
$$

We want to analyse the behaviour of our system near the equilibrium points, leading us to define the signed deviations:

$$
x(t)= N(t) - N^{*} \tag{3.2a}
$$

$$
y(t) = P(t) - P^{*} \tag{3.2b}
$$

We can rearrange to give:

$$
N(t) = x(t) + N^{*} \tag{3.3a}
$$

$$
P(t) =  y(t) + P^{*} \tag{3.3b}
$$

Differentiating $(3.3a)$ and $(3.3b)$ with respect to $t$ gives:

$$
\frac{dN}{dt} = \frac{dx}{dt} \tag{3.4a}
$$

$$
\frac{dP}{dt} = \frac{dy}{dt} \tag{3.4b}
$$

Using the equations in $(3.3)$ and in $(3.4)$ allow us to write our original set of equations from $(3.1)$ in terms of signed deviations from the equilibrium:

$$
\frac{dx}{dt} = F(x(t) + N^{*}, y(t) + P^{*}) \tag{3.5a}
$$

$$
\frac{dy}{dt} = G(x(t) + N^{*}, y(t) + P^{*}) \tag{3.5b}
$$

Our goal is to linearise the equations near the equilibrium point. To do this, we take Taylor expansions of the functions $F$ and $G$ about $(N^{*}, P^{*})$, and then retain their linear terms. This gives us an approximation of how small deviations around the equilibrium point change with time, giving us a solid foundation to analyse the local stability and classify the equilibrium.

Taking the Taylor expansion of $(3.5a)$ yields:

$$
F(x(t) + N^{*}, y(t) + P^{*}) = F(N^{*},P^{*}) + \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y + F_H(x,y) \tag{3.6}
$$

Where $F_H(x,y)$ is the higher order remainder. It approaches zero faster than the size of the displacement $\sqrt{x^2 + y^2}$ as $ (x,y) \rightarrow \boldsymbol{0}$, meaning it becomes negligible relative to the linear terms sufficiently close to the equilibrium. The rigorous justification is omitted here. 

We also should note that $F(N^{*},P^{*}) = 0$. We discard the higher-order remainder to obtain the linearised approximation. Applying the same method to $(3.5b)$ gives us the linearised system:

$$
\frac{dx}{dt} = \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y \tag{3.7a}
$$

$$
\frac{dy}{dt} = \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})}y \tag{3.7b}
$$

The equations in $(3.7)$ can be neatly expressed as:

$$
\begin{pmatrix} \frac{dx}{dt} \\ \frac{dy}{dt}\end{pmatrix} = \begin{pmatrix} \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})} \\ \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})} \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \boldsymbol{J}\boldsymbol{u} \tag{3.8}
$$

Where $\boldsymbol{u}(t) = \begin{pmatrix} x(t) \\ y(t) \end{pmatrix}$.

In technical terms, the matrix in $(3.8)$ is called the Jacobian matrix - called the community matrix in ecology. 

Using $(3.8)$ we look for equations of the form

$$
x(t) = x_0e^{\lambda t}, \quad y(t) = y_0e^{\lambda t} \tag{3.9}
$$

Where

$$
\boldsymbol{u}(t) = \begin{pmatrix} x(t) \\ y(t) \end{pmatrix} = e^{\lambda t}\begin{pmatrix} x_0 \\ y_0 \end{pmatrix} = e^{\lambda t} \boldsymbol{v} \tag{3.10}
$$

We call $\boldsymbol{u}(t)$ the eigenmode.

(Note: we look for equations in this form, because taking the derivative, substituting this and the original solution makes it so that the $e^{\lambda t}$ in every term cancels out, leaving us with an equation that is independent of $t$)

So with substitution of $(3.9)$ into $(3.7)$ we obtain:

$$
\lambda x_{0} = \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x_0 + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y_0 \tag{3.11a}
$$

$$
\lambda y_{0} = \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})}x_0 + \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})}y_0 \tag{3.11b}
$$

With some rearranging, $(3.11)$ can be neatly expressed as:

$$
\begin{pmatrix} \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})} - \lambda && \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})} \\ \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})} - \lambda \end{pmatrix} \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} = \boldsymbol{0} \tag{3.12}
$$

To make this not so messy looking, define $\boldsymbol{v} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix}$, we then can then simplify $(3.12)$ into:

$$
(\boldsymbol{J} - \lambda \boldsymbol{I})\boldsymbol{v} = \boldsymbol{0} \tag{3.13}
$$

A scalar $\lambda$ is called an eigenvalue of $\boldsymbol{J}$ if there exists a nonzero vector $\boldsymbol{v}$ satisfying (3.13). The vector $\boldsymbol{v}$ is then a corresponding eigenvector. The eigenvalues are of interest to us, so how do we extract them? Well, let's identify some facts about $(3.13)$, which will lead us to finding the eigenvalues.

We notice that algebraically that if $\boldsymbol{v} = \boldsymbol{0} \; \Rightarrow \; (\boldsymbol{J}- \lambda \boldsymbol{I})\boldsymbol{v}  = (\boldsymbol{J}- \lambda \boldsymbol{I})\boldsymbol{0} = \boldsymbol{0} \quad \forall \; \lambda \in \mathbb{C} $ which doesn't really help us mathematically.

Another way to notice this is that if we let $\boldsymbol{v} = \boldsymbol{0}$, then the eigenmode becomes $\boldsymbol{u}(t) = e^{\lambda t} \boldsymbol{0} = \boldsymbol{0}$, which means that $x(t) = 0$ and $y(t) = 0$ which means that $N(t) = N^{*}$ and $P(t) = P^{*}$, the equilibrium point! We want to explore what happens near the equilibrium point, not at it!

So, let us set $\boldsymbol{v} \neq \boldsymbol{0}$. Now, if $(\boldsymbol{J}- \lambda \boldsymbol{I})$ is invertible, that is, $\det(\boldsymbol{J}- \lambda \boldsymbol{I}) \neq 0$, we will have $\boldsymbol{v} = \boldsymbol{0}$, which contradicts our assumption. So $(\boldsymbol{J}- \lambda \boldsymbol{I})$ is non-invertible, that is $\det(\boldsymbol{J}- \lambda \boldsymbol{I}) = 0$. This gives us good ground to move on and find out the eigenvalues. We then expand the determinant to give us the characteristic equation:

$$
\lambda ^{2} - p\lambda + q = 0 \tag{3.14}
$$

Where $p = \operatorname{tr}(\boldsymbol{J})$ and $q = \det(\boldsymbol{J})$

We can use the quadratic formula to find the eigenvalues! But first, I will work through an example by finding the eigenvalues of a predator-prey model which is an extension of the Lotka-Volterra predator-prey model.
