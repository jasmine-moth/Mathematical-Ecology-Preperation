# Stability and Cycles

----

In the previous sections, we have discovered 2 different ecological models, with those being the logistic equation and the Lotka-Volterra model. Now we turn our attention to developing the rigour behind understanding stability and equilibrium points. 

Firstly, let us consider an autonomous predator-prey system:

$$
\frac{dN}{dt} = F(N,P) \tag{3.1a}
$$

$$
\frac{dP}{dt} = G(N,P) \tag{3.1b}
$$

Where $N$ is the population of prey, and $P$ is the predator population. We call them autonomous because the functions $F$ and $G$ have no explicit dependence on time. From earlier, we know that $(N^{*},P^{*})$ is an equilibrium point if:

$$
F(N^{*},P^{*}) = 0 \quad and \quad G(N^{*},P^{*}) = 0
$$

We want to analyse the behaviour of our system near the equilibrium points, leading us to define the signed deviations:

$$
x(t)= N(t) - N^{*} \tag{3.2a}
$$

$$
y(t) = P(t) - P^{*} \tag{3.2b}
$$

We can rearrange to give:

$$
N(t) = x(t) + N^{*} \tag{3.3a}
$$

$$
P(t) =  y(t) + P^{*} \tag{3.3b}
$$

Differentiating $(3.3a)$ and $(3.3b)$ with respect to $t$ gives:

$$
\frac{dN}{dt} = \frac{dx}{dt} \tag{3.4a}
$$

$$
\frac{dP}{dt} = \frac{dy}{dt} \tag{3.4b}
$$

Using the equations in $(3.3)$ and in $(3.4)$ allow us to write our original set of equations from $(3.1)$ with respect of distance from the equilibrium:

$$
\frac{dx}{dt} = F(x(t) + N^{*}, y(t) + P^{*}) \tag{3.5a}
$$

$$
\frac{dy}{dt} = G(x(t) + N^{*}, y(t) + P^{*}) \tag{3.5b}
$$

Our goal is to linearise the equations near the equilibrium point. To do this, we take Taylor expansion of the functions $F$ and $G$ about $(N^{*}, P^{*})$, and then retaining their linear terms. This gives us an approximation of how small deviations around the equilibrium point change with time, giving us a solid foundation to analyse the local stability and classifying the equilibrium.

Taking the Taylor expansion of $(3.5a)$ yields:

$$
F(x(t) + N^{*}, y(t) + P^{*}) = F(N^{*},P^{*}) + \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y + F_H(x,y) \tag{3.6}
$$

Where $F_H(x,y)$ are the higher order terms of the expansion. These terms can be disregarded because $F_H(x,y) \rightarrow 0$ as $ (x,y) \rightarrow 0$ faster compared to the linear terms. There is a rigorous way to show this, but for times sake, we omit this here.

We also should note that $F(N^{*},P^{*}) = 0$. We can apply the same methodology to $(3.5b)$ which gives us the pair of linearised equations:

$$
\frac{dx}{dt} = \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y \tag{3.7a}
$$

$$
\frac{dy}{dt} = \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})}y \tag{3.7b}
$$

The equations in $(3.7)$ can be neatly expressed as:

$$
\begin{pmatrix} \frac{dx}{dt} \\ \frac{dy}{dt}\end{pmatrix} = \begin{pmatrix} \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})} \\ \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})} \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \boldsymbol{J}\boldsymbol{x} \tag{3.8}
$$

In technical terms, the matrix in $(3.8)$ is called the Jacobian matrix - called the community matrix in ecology. 

Using $(3.8)$ we look for equations of the form

$$
x(t) = x_0e^{\lambda t}, \quad y(t) = y_0e^{\lambda t} \tag{3.9}
$$

(Note: we look for equations in this form, because taking the derivate, substituting this and the original solution makes it so that the $e^{\lambda t}$ in every term cancels out, leaving us with an equation that is independent of $t$)

So with substitution of $(3.9)$ into $(3.7)$ we obtain:

$$
\lambda x_{0} = \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x_0 + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y_0 \tag{3.10a}
$$

$$
\lambda y_{0} = \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})}x_0 + \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})}y_0 \tag{3.10b}
$$

With some rearranging, $(3.10)$ can be neatly expressed as:

$$
\begin{pmatrix} \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})} - \lambda && \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})} \\ \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})} - \lambda \end{pmatrix} \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} = \boldsymbol{0} \tag{3.11}
$$

To make this not so messy looking, define $\boldsymbol{v} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix}$, we then can then simplify $(3.11)$ into:

$$
\boldsymbol{v}(\boldsymbol{J} - \lambda \boldsymbol{I}) = \boldsymbol{0} \tag{3.12}
$$

We call $\lambda$ the eigenvalue and $\boldsymbol{v}$ the eigenvector.
