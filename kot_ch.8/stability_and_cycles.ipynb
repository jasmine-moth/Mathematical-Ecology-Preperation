# Stability and Cycles

----

In the previous sections, we have discovered 2 different ecological models, with those being the logistic equation and the Lotka-Volterra model. Now we turn our attention to developing the rigour behind understanding stability and equilibrium points. 

Firstly, let us consider an autonomous predator-prey system:

$$
\frac{dN}{dt} = F(N,P) \tag{3.1a}
$$

$$
\frac{dP}{dt} = G(N,P) \tag{3.1b}
$$

Where $N$ is the population of prey, and $P$ is the predator population. We call them autonomous because the functions $F$ and $G$ have no explicit dependence on time. From earlier, we know that $(N^{*},P^{*})$ is an equilibrium point if:

$$
F(N^{*},P^{*}) = 0 \quad and \quad G(N^{*},P^{*}) = 0
$$

We want to analyse the behaviour of our system near the equilibrium points, leading us to define the signed deviations:

$$
x(t)= N(t) - N^{*} \tag{3.2a}
$$

$$
y(t) = P(t) - P^{*} \tag{3.2b}
$$

We can rearrange to give:

$$
N(t) = x(t) + N^{*} \tag{3.3a}
$$

$$
P(t) =  y(t) + P^{*} \tag{3.3b}
$$

Differentiating $(3.3a)$ and $(3.3b)$ with respect to $t$ gives:

$$
\frac{dN}{dt} = \frac{dx}{dt} \tag{3.4a}
$$

$$
\frac{dP}{dt} = \frac{dy}{dt} \tag{3.4b}
$$

Using the equations in $(3.3)$ and in $(3.4)$ allow us to write our original set of equations from $(3.1)$ in terms of signed deviations from the equilibrium:

$$
\frac{dx}{dt} = F(x(t) + N^{*}, y(t) + P^{*}) \tag{3.5a}
$$

$$
\frac{dy}{dt} = G(x(t) + N^{*}, y(t) + P^{*}) \tag{3.5b}
$$

Our goal is to linearise the equations near the equilibrium point. To do this, we take Taylor expansions of the functions $F$ and $G$ about $(N^{*}, P^{*})$, and then retain their linear terms. This gives us an approximation of how small deviations around the equilibrium point change with time, giving us a solid foundation to analyse the local stability and classify the equilibrium.

Taking the Taylor expansion of $(3.5a)$ yields:

$$
F(x(t) + N^{*}, y(t) + P^{*}) = F(N^{*},P^{*}) + \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y + F_H(x,y) \tag{3.6}
$$

Where $F_H(x,y)$ is the higher order remainder. It approaches zero faster than the size of the displacement $\sqrt{x^2 + y^2}$ as $ (x,y) \rightarrow \boldsymbol{0}$, meaning it becomes negligible relative to the linear terms sufficiently close to the equilibrium. The rigorous justification is omitted here. 

We also should note that $F(N^{*},P^{*}) = 0$. We discard the higher-order remainder to obtain the linearised approximation. Applying the same method to $(3.5b)$ gives us the linearised system:

$$
\frac{dx}{dt} = \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y \tag{3.7a}
$$

$$
\frac{dy}{dt} = \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})}x + \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})}y \tag{3.7b}
$$

The equations in $(3.7)$ can be neatly expressed as:

$$
\begin{pmatrix} \frac{dx}{dt} \\ \frac{dy}{dt}\end{pmatrix} = \begin{pmatrix} \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})} \\ \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})} \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \boldsymbol{J}\boldsymbol{u} \tag{3.8}
$$

Where $\boldsymbol{u}(t) = \begin{pmatrix} x(t) \\ y(t) \end{pmatrix}$.

In technical terms, the matrix in $(3.8)$ is called the Jacobian matrix - called the community matrix in ecology. 

Using $(3.8)$ we look for equations of the form

$$
x(t) = x_0e^{\lambda t}, \quad y(t) = y_0e^{\lambda t} \tag{3.9}
$$

Provided that the eigenvectors $\boldsymbol{v}_1$ and $\boldsymbol{v}_2$ are linearly independent, the general solution is given by:

$$
\boldsymbol{u}(t) = \sum_{i=1}^{2}{c_i\boldsymbol{u}_i} = \sum_{i=1}^{2} c_i e^{\lambda_i t} \boldsymbol{v}_i \tag{3.10}
$$

We call $\boldsymbol{u}_i(t) = e^{\lambda_i t} \boldsymbol{v}_i$ the eigenmode.

(Note: we look for equations in this form, because taking the derivative, substituting this and the original solution makes it so that the $e^{\lambda t}$ in every term cancels out, leaving us with an equation that is independent of $t$)

So with substitution of $(3.9)$ into $(3.7)$ we obtain:

$$
\lambda x_{0} = \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})}x_0 + \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})}y_0 \tag{3.11a}
$$

$$
\lambda y_{0} = \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})}x_0 + \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})}y_0 \tag{3.11b}
$$

With some rearranging, $(3.11)$ can be neatly expressed as:

$$
\begin{pmatrix} \left.\frac{\partial{F}}{\partial{N}}\right|_{(N^{*},P^{*})} - \lambda && \left.\frac{\partial{F}}{\partial{P}}\right|_{(N^{*},P^{*})} \\ \left.\frac{\partial{G}}{\partial{N}}\right|_{(N^{*},P^{*})} && \left.\frac{\partial{G}}{\partial{P}}\right|_{(N^{*},P^{*})} - \lambda \end{pmatrix} \begin{pmatrix} x_0 \\ y_0 \end{pmatrix} = \boldsymbol{0} \tag{3.12}
$$

To make this not so messy looking, define $\boldsymbol{v} = \begin{pmatrix} x_0 \\ y_0 \end{pmatrix}$, we then can then simplify $(3.12)$ into:

$$
(\boldsymbol{J} - \lambda \boldsymbol{I})\boldsymbol{v} = \boldsymbol{0} \tag{3.13}
$$

A scalar $\lambda$ is called an eigenvalue of $\boldsymbol{J}$ if there exists a nonzero vector $\boldsymbol{v}$ satisfying (3.13). The vector $\boldsymbol{v}$ is then a corresponding eigenvector. The eigenvalues are of interest to us, so how do we extract them? Well, let's identify some facts about $(3.13)$, which will lead us to finding the eigenvalues.

We notice that algebraically that if $\boldsymbol{v} = \boldsymbol{0} \; \Rightarrow \; (\boldsymbol{J}- \lambda \boldsymbol{I})\boldsymbol{v}  = (\boldsymbol{J}- \lambda \boldsymbol{I})\boldsymbol{0} = \boldsymbol{0} \quad \forall \; \lambda \in \mathbb{C} $ which doesn't really help us mathematically.

Another way to notice this is that if we let $\boldsymbol{v} = \boldsymbol{0}$, then the eigenmode becomes $\boldsymbol{u}(t) = e^{\lambda t} \boldsymbol{0} = \boldsymbol{0}$, which means that $x(t) = 0$ and $y(t) = 0$ which means that $N(t) = N^{*}$ and $P(t) = P^{*}$, the equilibrium point! We want to explore what happens near the equilibrium point, not at it!

So, let us set $\boldsymbol{v} \neq \boldsymbol{0}$. Now, if $(\boldsymbol{J}- \lambda \boldsymbol{I})$ is invertible, that is, $\det(\boldsymbol{J}- \lambda \boldsymbol{I}) \neq 0$, we will have $\boldsymbol{v} = \boldsymbol{0}$, which contradicts our assumption. So $(\boldsymbol{J}- \lambda \boldsymbol{I})$ is non-invertible, that is $\det(\boldsymbol{J}- \lambda \boldsymbol{I}) = 0$. This gives us good ground to move on and find out the eigenvalues. We then expand the determinant to give us the characteristic equation:

$$
\lambda ^{2} - p\lambda + q = 0 \tag{3.14}
$$

Where $p = \operatorname{tr}(\boldsymbol{J})$ and $q = \det(\boldsymbol{J})$

We can use the quadratic formula to find the eigenvalues! But first, I will work through an example by finding the eigenvalues of a predator-prey model which is an extension of the Lotka-Volterra predator-prey model.

## Finding the Eigenvalues of The Rosenzweig-MacArthur Predator-Prey Model

In this example, I will be working through a specially parameterised Rosenzweig-MacArthur predator-prey model. I will give the general form, and then state the parameters I will be using:

$$
\frac{dN}{dt} = rN\left( 1 - \frac{N}{K}\right) - \frac{aNP}{1+ahN} \tag{3.15a}
$$

$$
\frac{dP}{dt} = P\left (\frac{eaN}{1+ahN} - m\right) \tag{3.15b}
$$

Where:

- $N$: population of prey.
- $P$: population of predators.
- $t$: time.
- $r$: intrinsic per-capita growth rate of prey.
- $K$: the prey carrying capacity in the absence of predators.
- $a$: predator attack or encounter-rate coefficient.
- $h$: handling time per captured prey.
- $e$: conversion efficiency - the contribution of consumed prey to predator recruitment.
- $m$: per-capita predator mortality rate.

In the example I will be working through, the following specific parameterised version where: $r=1, \ K=2, \ a=1, \ h=1,\ e=1,\ m=\frac{1}{2}$. This gives us the autonomous predator-prey system:

$$
\frac{dN}{dt} = F(N,P) = N\left( 1 - \frac{N}{2}\right) - \frac{NP}{1+N} \tag{3.16a}
$$

$$
\frac{dP}{dt} = G(N,P) = P\left( \frac{N}{1+N} - \frac{1}{2}\right) \tag{3.16b}
$$

We need to construct the Jacobian matrix for this system, thus we need to find the partial derivatives of $F$ and $G$ with respect to $N$ and $P$, which are:

$$
\frac{\partial{F}}{\partial{N}} = 1 - N - \frac{P}{(1+N)^2} \tag{3.17a}
$$
$$
\frac{\partial{F}}{\partial{P}} = - \frac{N}{1+ N} \tag{3.17b}
$$
$$
\frac{\partial{G}}{\partial{N}} = \frac{P}{(1+N)^2} \tag{3.17c}
$$
$$
\frac{\partial{G}}{\partial{P}} = \frac{N}{1+N} - \frac{1}{2} \tag{3.17d}
$$

Thus, using the partial derivatives from $(3.17)$ gives us the Jacobian matrix:

$$
\boldsymbol{J}(N,P) = \begin{pmatrix} 1 - N - \frac{P}{(1+N)^2} & - \frac{N}{1+ N} \\ \frac{P}{(1+N)^2} & \frac{N}{1+N} - \frac{1}{2} \end{pmatrix} \tag{3.18}
$$

The trace of $\boldsymbol{J}(N,P)$ is:

$$
\operatorname{tr}(\boldsymbol{J}(N,P)) = \frac{1}{2} - \frac{N^2}{1+N} - \frac{P}{(1+N)^2} \tag{3.19}
$$

And the determinant of $\boldsymbol{J}(N,P)$ is:

$$
\det(\boldsymbol{J}(N,P)) = \frac{P - (N-1)^2(1 +N)}{2(1+N)^2} \tag{3.20}
$$

Thus, using $(3.14)$ the characteristic equation for the system in terms of $N$ and $P$ is

$$
\lambda ^2 - \left(\frac{1}{2} - \frac{N^2}{1+N} - \frac{P}{(1+N)^2}\right)\lambda + \frac{P - (N-1)^2(1 +N)}{2(1+N)^2} = 0 \tag{3.21}
$$

Now, we need to find the equilibrium points so we can substitute $E^* = (N^*, P^*)$ into $(3.21)$ in order to find the eigenvalues of $\boldsymbol{J}(E^*)$. Recall that the equilibrium point is given when $F(N^*, P^*) = 0$ and $G(N^*, P^*) = 0$.

It is straightforward that $(0,0)$ is an equilibrium point, which signifies extinction of both predator and prey.

Setting $P=0$ gives $N=0$ or $N=2$. Setting $N=0$ gives $P=0$, so there are no additional equilibria on this axis, whereas $N=2$ gives the predator-free equilibrium $(2,0)$, where the prey are at carrying capacity $K$.

To find the coexistence equilibrium, we assume $N>0$ and $P>0$ and solve $F(N^*,P^*) = 0$ and $G(N^*,P^*) = 0$ simultaneously using substitution. This gives $N=1$, and substituting this into either of the nullcline equations gives $P=1$, thus giving us the coexistence equilibrium $(1,1)$. No additional equilibria arise when $N=0$.

We now have established the 3 equilibrium points, namely $(0,0)$, $(2,0)$, and $(1,1)$. Meaning we can evaluate the characteristic equation produced in $(3.21)$ at each equilibrium point to obtain the eigenvalues for $\boldsymbol{J}(N^*,P^*)$, giving:

$$
E_1 = (0,0): \quad \lambda_1 = 1, \quad \lambda_2 = -\frac{1}{2} \tag{3.22a}
$$
$$
E_2 = (2,0): \quad \lambda_1 = -1, \quad \lambda_2 = \frac{1}{6} \tag{3.22b}
$$
$$
E_3 = (1,1): \quad \lambda_{1,2} = -\frac{1}{8} \pm \frac{\sqrt{7}}{8}i \tag{3.22c}
$$
---

After indulging in eigenvalue calculations, we find ourselves asking how they are of use in such autonomous predator-prey systems. The short answer is that the eigenvalues of a linearised system enable us to provide a rigorous way to classify the type and local stability of an equilibrium point. For example, is it stable or unstable? Is it a node, a saddle point, a center or a focus? 

To do this, we analyse the eigenvalues' real and imaginary parts, and can use this to classify the equilibria of a linearised system. I will state claims about the eigenvalues, how they classify the equilibrium point, and then give a justification of why we can make generalised classifications. I should mention that an equilibrium point is hyperbolic if every single eigenvalue of its Jacobian has a non-zero real part. For a continuously differentiable system, hyperbolicity ensures that the local stability conditions from the linearised system carry over to the nonlinear system. If an eigenvalue contains a zero real part, further analysis may be required where we investigate how the nonlinear terms behave near the equilibrium point. 

Let us begin:

##### 1. If both eigenvalues are distinct, real and negative - the equilibrium is a stable node.

So we have $\lambda_1, \lambda_2 \in \mathbb{R}^-$, with $\lambda_1 \neq \lambda_2$. We assume them to be distinct in this explanation because distinct eigenvalues guarantee linearly independent eigenvectors, allowing us to use $(3.10)$. If the eigenvalues were equal, two linearly independent eigenvectors may not exist.

We recall that from $(3.10)$ we have $\boldsymbol{u}_i(t) = e^{\lambda_i t}\boldsymbol{v}_i$ where $i = 1, 2$. Since $\lambda_i < 0$ for $i = 1, 2$, we see that $\lambda_i t \rightarrow - \infty$ as $t \rightarrow \infty \Rightarrow e^{\lambda_i t} \rightarrow 0$ as $t \rightarrow \infty $. Since each $\boldsymbol{v}_i$ is constant, it follows that $\boldsymbol{u}_i(t) \rightarrow \boldsymbol0$ for $i=1,2$ as $t \rightarrow \infty \Rightarrow \boldsymbol{u}(t) \rightarrow \boldsymbol0$ as $t \rightarrow \infty$. This behaviour indicates that this is an asymptotically stable node, the negative eigenvalues give decay, and since they are real and distinct,the equilibrium is classified as a node.

##### 2. If both eigenvalues are distinct, real and positive - the equilibrium is an unstable node.

We have $\lambda_1, \lambda_2 \in \mathbb{R}^+$, from $(3.10)$ we have $\boldsymbol{u}_i(t) = e^{\lambda_i t}\boldsymbol{v}_i$ where $i = 1, 2$. Since $\lambda_i > 0$ for $i = 1, 2$, we see that $\lambda_i t \rightarrow  \infty$ as $t \rightarrow \infty \Rightarrow e^{\lambda_i t} \rightarrow \infty $ as $t \rightarrow \infty $. Since each $\boldsymbol{v}_i$ is constant and non-zero, we see that the norm of the eigenmode grows without bound: $\|\boldsymbol u_i(t)\| = e^{\lambda_i t}\|\boldsymbol v_i\| \rightarrow \infty$ as $ t\to\infty$. Because the eigenvectors are linearly independent, every non-zero initial deviation satisfies $\|\boldsymbol u(t)\|\rightarrow\infty$ as $t\to\infty$. This behaviour indicates that this is an unstable node, the positive eigenvalues give growth, and since the eigenvalues are real and distinct, the equilibrium is classified as a node. We note here that the unbounded growth only applies to the linearised system. For the original nonlinear system, this represents local instability, and does not necessarily imply unbounded growth everywhere.

##### 3. If both eigenvalues are distinct real, but of opposing signs - the equilibrium is a saddle point.

Choose $\lambda_1 > 0, \lambda_2 < 0$ without loss of generality with $\lambda_1, \lambda_2 \in \mathbb{R}$. The eigenmodes will be $\boldsymbol{u}_1 = e^{\lambda_1 t}\boldsymbol{v}_1$ and $\boldsymbol{u}_2 = e^{\lambda_2 t}\boldsymbol{v}_2$. For $\boldsymbol{u}_1$ as $t\to\infty$, $\|\boldsymbol{u}_1\| \to \infty$, showing it is unbounded as $t$ increases. For $\boldsymbol{u}_2$ as $t\to\infty, \boldsymbol{u}_2 \to \boldsymbol0$ showing that it decreases to zero as $t$ increases. For the eigenvector direction in $\boldsymbol{u}_2$, trajectories along $\boldsymbol{v}_2$ approach the equilibrium. Whereas for $\boldsymbol{u}_1$, trajectories along $\boldsymbol{v}_1$ move away from the equilibrium. This is what we call a saddle point. It is unstable: any initial deviation with $c_1 \neq 0 $ eventually grows without bound in the linearised system. However, if $c_1 = 0$, then the solution would approach the equilibrium.

##### 4.If eigenvalues are complex with a negative real part - the equilibrium is a stable focus.

We have $\lambda_{\pm} = \alpha \pm \beta i$ where $\alpha \in \mathbb{R}^-$ and $\beta \in \mathbb{R}^+$. We then use Euler's formula to rewrite the exponential factor in each eigenmode to obtain $e^{(\alpha\pm i\beta)t} = e^{\alpha t}\bigl(\cos(\beta t)\pm i\sin(\beta t)\bigr)$. Using this, we can see that the factor $\alpha$ controls the amplitude of each eigenmode, for $\alpha < 0$ it is decay. $\beta$ contributes to oscillation of the eigenmode. Combining the conjugate eigenmodes gives real trajectories that spiral inwards to the equilibrium: an asymptotically stable focus. 

##### 5.If eigenvalues are complex with a positive real part - the equilibrium is an unstable focus.

We have $\lambda_{\pm} = \alpha \pm \beta i$ where $\alpha \in \mathbb{R}^+$ and $\beta \in \mathbb{R}^+$. Again, use Euler's formula to rewrite the exponential factor in each eigenmode to obtain $e^{(\alpha\pm i\beta)t} = e^{\alpha t}\bigl(\cos(\beta t)\pm i\sin(\beta t)\bigr)$. As before, the factor $\alpha$ controls the amplitude of each eigenmode, but now $\alpha > 0$, so the amplitude grows without bound in the linearised system. $\beta$ contributes to oscillation of the eigenmode. For the original non-linear system, we have local outward spiralling and instability at the equilibrium.

##### 6. If eigenvalues are nonzero and purely imaginary, the linearised system has a center, while the nonlinear system may have a center, a stable focus or an unstable focus.

We have $\lambda_{\pm} = \alpha \pm \beta i$ where $\alpha = 0$ and $\beta \in \mathbb{R}^+$. Thus, there is no exponential growth or decay. The real solutions in the linearised system oscillate in a closed ellipse around the equilibrium. It is a stable center, but not asymptotically stable as it does not converge to the equilibrium. The equilibrium is not hyperbolic since hyperbolicity requires every eigenvalue to have nonzero real parts. Although the nonlinear terms of the system are small near the equilibrium, there is no exponential factor that contributes to decay or growth in the linear system, which usually dominates the effects. This means that effects of nonlinear terms can accumulate over time, and could contribute to outward or inward spiralling or preserve closed orbits. Further analysis is needed, such as a conserved quantity or a calculation for how distance from the equilibrium changes.

## Classifying the Equilibria of The Simplified Rosenzweig-MacArthur Predator-Prey Model

We have now developed to right tools to continue our investigation from earlier. 

From $(3.22a)$ the equilibrium $E_1 = (0,0)$ has eigenvalues $\lambda_1 = 1$ and $\lambda_2 = -\frac{1}{2}$. The eigenvalues are both real but of opposite sign, meaning that $E_1$ is an unstable saddle point. We know that $E_1$ is the extinction of both predator and prey, but the saddle point tells us that without prey the predators will move towards extinction. However, if a small number of prey were introduced in the absence of predators, the prey population will grow, moving the system away from the total-extinction equilibrium.

From $(3.22b)$ the equilibrium $E_2 = (2,0)$ has eigenvalues $\lambda_1 = -1$ and $\lambda_2 = \frac{1}{6}$. The eigenvalues are both real but of opposite sign, meaning that $E_2$ is an unstable saddle point. Without predators, the prey are already at their carrying capacity of $K=2$, meaning a small change in prey abundance will return back to a population of $2$. With a small introduction of predators, predators initially increase because the population of the prey supports positive predator growth. This moves the system away from the equilibrium point, and predation causes prey to initially decrease. 

From $(3.22c)$ the coexistence equilibrium $E_3 = (1,1)$ has eigenvalues $\lambda_{1,2} = -\frac{1}{8} \pm \frac{\sqrt{7}}{8}i$. The eigenvalues are complex with a negative real part, meaning that $E_3$ is a stable focus. If the predator and prey stay at this point, there will be no change in the system. If the predator and prey are sufficiently close to the equilibrium, they will undergo damped oscillations, approaching the coexistence equilibrium as time increases.

We can use our results to predict what the directional field near each equilibria looks like. I will now plot this, so we can compare the plotted behaviour with our predictions.
